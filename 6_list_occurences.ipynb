```python
# Read number of elements
n = int(input("Enter number of elements: "))

# Read list elements
lst = []
print("Enter the elements (e.g., 12,19,5,5,6):") # Clarify expected format
elements_str = input() # Read the whole line
elements_str_list = elements_str.split(',') # Split by comma
for s in elements_str_list:
    lst.append(int(s.strip())) # Convert each part to int

# Count occurrences
count_19 = lst.count(19)
count_5 = lst.count(5)

# Check condition
if count_19 == 2 and count_5 >= 3:
    print("Condition satisfied")
else:
    print("Condition not satisfied")
```
