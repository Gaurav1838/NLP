##1. porter's stemmer

```python
from nltk.stem import PorterStemmer

# Create stemmer object
ps = PorterStemmer()

# Input sentence
text = input("Enter a sentence:\n")

# Split into words
words = text.split()

print("\nPorter Stemmer Output:")
for word in words:
    print(word, "→", ps.stem(word))
```

##2. snowball stemmer

```python
from nltk.stem import SnowballStemmer

# Create Snowball stemmer for English
stemmer = SnowballStemmer("english")

# Input sentence
text = input("Enter a sentence:\n")

# Split into words
words = text.split()

print("\nSnowball Stemmer Output:")
for word in words:
    print(word, "→", stemmer.stem(word))
```

##3.  lancaster stemmer

```python
from nltk.stem import LancasterStemmer

# Create Lancaster stemmer
stemmer = LancasterStemmer()

# Input sentence
text = input("Enter a sentence:\n")

# Split into words
words = text.split()

print("\nLancaster Stemmer Output:")
for word in words:
    print(word, "→", stemmer.stem(word))
```

##4. Regexp Stemmer

```python
from nltk.stem import RegexpStemmer

# Create Regexp stemmer
# This removes common suffixes like ing, ed, s
stemmer = RegexpStemmer('ing$|ed$|s$')

# Input sentence
text = input("Enter a sentence:\n")

# Split into words
words = text.split()

print("\nRegexp Stemmer Output:")
for word in words:
    print(word, "→", stemmer.stem(word))
```
