```python
# Read input strings
s1 = input("Enter source string: ")
s2 = input("Enter target string: ")

m = len(s1)
n = len(s2)

# Create DP table
dp = [[0 for j in range(n + 1)] for i in range(m + 1)]

# Initialize first row and column
for i in range(m + 1):
    dp[i][0] = i       # deletion
for j in range(n + 1):
    dp[0][j] = j       # insertion

# Fill DP table
for i in range(1, m + 1):
    for j in range(1, n + 1):
        if s1[i - 1] == s2[j - 1]:
            dp[i][j] = dp[i - 1][j - 1]
        else:
            dp[i][j] = 1 + min(
                dp[i - 1][j],      # deletion
                dp[i][j - 1],      # insertion
                dp[i - 1][j - 1]   # substitution
            )

# Output result
print("Minimum Edit Distance:", dp[m][n])
```
