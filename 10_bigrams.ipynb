```python
# Collection of 3 documents
documents = [
    "data science is fun",
    "data science is useful",
    "science is fun and useful"
]

# List to store all words
words = []

# Combine all documents
for doc in documents:
    words.extend(doc.split())

# Generate bi-grams
bigrams = []
for i in range(len(words) - 1):
    bigrams.append((words[i], words[i + 1]))

# Count frequency of bi-grams
freq = {}
for bg in bigrams:
    freq[bg] = freq.get(bg, 0) + 1

# Total number of bi-grams
print("Total number of bi-grams:", len(bigrams))

# Sort bi-grams by frequency
sorted_bigrams = sorted(freq.items(), key=lambda x: x[1], reverse=True)

# Display top 5 bi-grams
print("\nTop 5 most common bi-grams:")
for bg, count in sorted_bigrams[:5]:
    print(bg, ":", count)
```
