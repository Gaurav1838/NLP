```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

# Download required data (run once)
nltk.download('punkt')

# Given corpus
corpus = "data science is fun data science is useful"

# Input sentence
sentence = input("Enter a sentence:\n")

# Tokenize corpus and sentence
corpus_words = word_tokenize(corpus.lower())
sentence_words = word_tokenize(sentence.lower())

# Frequency distribution of corpus words
fdist = FreqDist(corpus_words)

# Total number of words in corpus
total_words = len(corpus_words)

# Calculate sentence probability
probability = 1
for word in sentence_words:
    probability *= fdist[word] / total_words

# Output
print("Probability of the sentence:", probability)
```
