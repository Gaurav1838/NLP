## From User

```python
paragraph = input("Enter a paragraph:\n")
```

```python
sentences = paragraph.split('.')

print("\nSentences from user input:")
for s in sentences:
    if s.strip() != "":
        print("-", s.strip())
```

```python
words = paragraph.split()
```

```python
print("\nWords from user input:")
print(words)
```

## from file

```python
file = open("input.txt", "r")
file_paragraph = file.read()
file.close()
```

```python
file_sentences = file_paragraph.split('.')
```

```python
print("\nSentences from file:")
for s in file_sentences:
    if s.strip() != "":
        print("-", s.strip())
```

```python
file_words = file_paragraph.split()

print("\nWords from file:")
print(file_words)
```
